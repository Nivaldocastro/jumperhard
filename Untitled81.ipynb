```python
import pandas as pd
import matplotlib.pyplot as plt

# Ler o dataset
df = pd.read_csv("metricas_treinamento_2.csv")

# Criar o gráfico
plt.figure(figsize=(14, 7))

# Lista de ambientes
ambientes = df["ambiente"].unique()

# Criar uma linha para cada ambiente
for ambiente in ambientes:
    dados_ambiente = df[df["ambiente"] == ambiente]

    plt.plot(
        dados_ambiente["episodio"],
        dados_ambiente["plataformas_alcancadas"],
        marker="o",
        markersize=2,
        linewidth=1.5,
        label=ambiente
    )

# Configurações
plt.title("Plataformas Alcançadas por Ambiente")
plt.xlabel("Episódio")
plt.ylabel("Plataformas Alcançadas")

plt.legend(
    title="Ambiente",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# Criar uma figura com 16 gráficos
fig, axes = plt.subplots(4, 4, figsize=(16, 12))

# Transformar a matriz de eixos em uma lista
axes = axes.flatten()

for i, ambiente in enumerate(ambientes):

    dados = df[df["ambiente"] == ambiente]

    axes[i].plot(
        dados["episodio"],
        dados["plataformas_alcancadas"],
        marker="o",
        markersize=3,
        linewidth=1.5
    )

    axes[i].set_title(ambiente)
    axes[i].set_xlabel("Episódio")
    axes[i].set_ylabel("Plataformas")

    axes[i].grid(True, alpha=0.3)

# Ajustar espaçamento
plt.suptitle(
    "Plataformas Alcançadas por Ambiente",
    fontsize=18
)

plt.tight_layout()

plt.show()
```

```python
# Criar tabela
tabela = df.pivot_table(
    index="ambiente",
    columns="episodio",
    values="plataformas_alcancadas",
    aggfunc="mean"
)

# Criar o gráfico
plt.figure(figsize=(16, 8))

sns.heatmap(
    tabela,
    cmap="viridis",
    annot=False,
    linewidths=0.2,
    cbar_kws={
        "label": "Plataformas alcançadas"
    }
)

# Mostrar apenas alguns episódios no eixo X
episodios = tabela.columns

# Selecionar episódios de 5 em 5
ticks = [
    i for i, episodio in enumerate(episodios)
    if episodio % 5 == 0 or episodio == 1 or episodio == 67
]

labels = [
    episodios[i]
    for i in ticks
]

plt.xticks(
    ticks=[i + 0.5 for i in ticks],
    labels=labels,
    rotation=0
)

plt.title(
    "Plataformas Alcançadas por Ambiente e Episódio",
    fontsize=16
)

plt.xlabel("Episódio")
plt.ylabel("Ambiente")

plt.tight_layout()

plt.show()
```

```python
df = pd.read_csv("metricas_treinamento_2.csv")

df["media_movel"] = (
    df.groupby("ambiente")["plataformas_alcancadas"]
      .transform(lambda x: x.rolling(10, min_periods=1).mean())
)

ambientes = df["ambiente"].unique()

fig, axes = plt.subplots(4, 4, figsize=(16, 12))
axes = axes.flatten()

for i, ambiente in enumerate(ambientes):

    dados = df[df["ambiente"] == ambiente]

    axes[i].plot(
        dados["episodio"],
        dados["media_movel"],
        linewidth=2
    )

    axes[i].set_title(ambiente)
    axes[i].set_xlabel("Episódio")
    axes[i].set_ylabel("Média")

    axes[i].grid(True, alpha=0.3)

plt.suptitle(
    "Tendência das Plataformas Alcançadas",
    fontsize=18
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("metricas_treinamento_3.csv")

plt.figure(figsize=(16, 7))

sns.boxplot(
    data=df,
    x="ambiente",
    y="plataformas_alcancadas"
)

plt.title("Distribuição de Plataformas Alcançadas por Ambiente")
plt.xlabel("Ambiente")
plt.ylabel("Plataformas Alcançadas")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
media = (
    df.groupby("ambiente")["plataformas_alcancadas"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(14, 7))

media.plot(kind="bar")

plt.title("Média de Plataformas Alcançadas por Ambiente")
plt.xlabel("Ambiente")
plt.ylabel("Média de Plataformas")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["plataformas_alcancadas"],
    bins=20
)

plt.title("Distribuição de Plataformas Alcançadas")
plt.xlabel("Plataformas Alcançadas")
plt.ylabel("Frequência")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
variaveis = [
    "episodio",
    "plataformas_alcancadas",
    "tempo_sobrevivido",
    "pulos_realizados"
]

# Calcular a matriz de correlação
correlacao = df[variaveis].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlacao,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5
)

plt.title("Matriz de Correlação das Métricas do Treinamento")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="tempo_sobrevivido",
    y="plataformas_alcancadas",
    hue="ambiente",
    alpha=0.7
)

plt.title("Relação entre Tempo de Sobrevivência e Plataformas Alcançadas")
plt.xlabel("Tempo Sobrevivido (s)")
plt.ylabel("Plataformas Alcançadas")

plt.tight_layout()
plt.show()
```
